## Opbygning, redigering og navigation

En Jupyter Notebook består som sagt af hhv. Markdown- og kodefelter. Man kan oprette, redigere, fjerne og flytte rundt på felter som man har lyst til.

I kodefelterne skriver man - sjovt nok, - kode i form af Python. Kodefelterne eksekveres hver for sig, men alle felter kan også eksekveres én efter én ved trykke på `Run All/Kør alle`-knappen i toppen af VS-Code.

Modsat en alm. Python-fil, skal man være lidt opmærksom på eksekveringsrækkefølgen; Har man kørt et felt, skal man f.eks. være opmærksom på, at variabler i dette felt nu også er gældende i de resterende felter (både over/under) - hvorfor det er smart at kende til `Run All/Kør alle`-knappen i toppen, således at alle felter (og dermed også variabler) eksekveres kronologisk.

I Markdown-felterne skriver man naturligvis Markdown --- mere om dette senere.

Når man har markeret/"er i et" felt, kan man befinde sig i to tilstande; ***Command Mode*** og ***Edit Mode***.

***Command Mode*** kan du bruge tastaturgenveje til at manipulere dokumentet og samtidig vil f.eks. Markdown også være formateret pænt. Man bruger ofte denne tilstand når man skal "se/kigge på" feltet.

***Edit Mode*** bruges når man redigere et felts indhold --- derfor vil man også se den "rå" Markdown-kode og ikke den pæne og formaterede version. Prøv f.eks. at trykke `ENTER` eller dobbeltklikke på dette felt. For at komme tilbage, tryk på fluebenet i feltets højre øverste hjørne eller tryk `ESC`.

## Introduktion til Markdown

Nu har vi nævnt Markdown en del gange, men endnu ikke præciseret eller forklaret hvad det er.

Markdown er en måde at formatere tekst på. I Word eller Google Docs er man vant til at bruge knapper til at gøre ting fede, ændre skriftstørrelse, etc. I Markdown skriver man det direkte sammen med indholdet/teksten --- hvorfor det kaldes et Markup-sprog. Derfor skal Markdown også vises i et miljø der kan oversætte Markdown --- f.eks. en Jupyter Notebook. Man kan også skrive rene Markdown-filer, disse ender på `.md` --- Markdown kan også skrives i f.eks. Apple Notes eller Obsidian.

Overskrifter markeres med `#`. Et enkelt `#` betyder den største overskrift. 2 `#` den næst største osv.
# Heading 1
## Heading 2
### Heading 3
#### Heading 4

Tekst kan skrives `italic` ved at omklamre dem med `*`, altså bliver `*Slagelse*` til *Slagelse*.
Ved `bold` bruger man `**`, altså bliver `**Slagelse**` til **Slagelse**.
`bold` og `italic` benytter man blot tre, altså bliver `***Slagelse***` til ***Slagelse***.

Man kan tilføje et kode lignende felt til tekst som her: `Hej` ved at skrive \`Hej\`. Man kan også tilføje skjulte kommentarer ved at skrive \<!--Tekst-->. <!--Kommentar-->

Dertil kan man tilføje links i sin Markdown tekst ved at skrive \[Hyperlink tekst\]\(URL\). F.eks. ved at klikke på følgende "her": [her](https://jupyter-tutorial.readthedocs.io/en/24.1.0/notebook/shortcuts.html), så følger man linket til tastaturgenvejende.

Se følgende link for resterende syntax regler: [her](https://www.markdownguide.org/cheat-sheet/).

**Protip:** Du kan skrive LaTeX direkete i markdown ved at bruge `$` --- `$<LaTeX på en linje>$`, `$$<LaTeX på tværs af flere linjer>$$`.

**Protip:** Markdown er meget mere token efficient end f.eks. PDF'er, når du skal have diverse LLM'er til at lave dine ting.
